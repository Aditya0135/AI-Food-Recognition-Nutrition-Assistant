## Function: Search Food Recipe (ONE API CALL)

```python
def search_food_recipe(food_name: str):
    """
    Search for food recipe with EVERYTHING in ONE API call.
    """
    try:
        url = f"{API_BASE_URL}/recipes/complexSearch"
        params = {
            "query": food_name,
            "number": 1,
            "ranking": "popularity",
            "apiKey": "a03c1ff555034e0aaf14a34ba5d93d92",
            "addRecipeNutrition": True,
        }

        response = requests.get(url, params=params, timeout=5)
        response.raise_for_status()

        data = response.json()
        return data["results"][0]

    except Exception as e:
        print(f"❌ Recipe search error: {str(e)}")
        return None

search_food_recipe("pizza")
```
